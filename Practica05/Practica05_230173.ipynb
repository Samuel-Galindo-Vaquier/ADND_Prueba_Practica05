```python
def bar_hor(df, col, title, color, w=None, h=None, lm=0, limit=100, return_trace=False, rev=False):
    cnt_srs = df[col].value_counts()
    yy = cnt_srs.head(limit).index[::-1]
    xx = cnt_srs.head(limit).values[::-1]

    if rev:
        yy = cnt_srs.tail(limit).index[::-1]
        xx = cnt_srs.tail(limit).values[::-1]

    trace = go.Bar(y=yy, x=xx, orientation='h',
                   marker=dict(color=color))

    if return_trace:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


def bar_hor_noagg(x, y, title, color, w=None, h=None, lm=0, limit=100, rt=False):
    trace = go.Bar(y=x, x=y, orientation='h',
                   marker=dict(color=color))

    if rt:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


def bar_ver_noagg(x, y, title, color, w=None, h=None, lm=0, rt=False):
    trace = go.Bar(y=y, x=x,
                   marker=dict(color=color))

    if rt:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


trace1 = bar_hor(
    donors_df,
    'Donor City',
    "Ciudades con el mayor número de donantes",
    '#c2d2ed',
    500, 400, 200,
    limit=10,
    return_trace=True
)

trace2 = bar_hor(
    donors_df,
    'Donor State',
    "Estados con el mayor número de donantes",
    '#c2d2ed',
    500, 400, 200,
    limit=10,
    return_trace=True
)

trace4 = bar_hor(
    donors_df,
    'Donor State',
    "Estados con el menor número de donantes",
    '#c2d2ed',
    600, 400, 200,
    limit=10,
    return_trace=True,
    rev=True
)

fig = tools.make_subplots(
    rows=1,
    cols=2,
    print_grid=False,
    subplot_titles=[
        'Estados con el mayor número de donantes',
        'Estados con el menor número de donantes'
    ]
)

fig.append_trace(trace2, 1, 1)
fig.append_trace(trace4, 1, 2)

fig['layout'].update(height=400, showlegend=False)
iplot(fig)
```

```python
# Crear variables de fecha para los proyectos
projects_df['Posted Date'] = pd.to_datetime(projects_df['Project Posted Date'])

projects_df['Posted Year'] = projects_df['Posted Date'].dt.year
projects_df['Posted Month'] = projects_df['Posted Date'].dt.month
projects_df['Posted Day'] = projects_df['Posted Date'].dt.weekday
projects_df['Posted Quarter'] = projects_df['Posted Date'].dt.quarter

projects_df['Funded Date'] = pd.to_datetime(projects_df['Project Fully Funded Date'])
projects_df['Funded Year'] = projects_df['Funded Date'].dt.year
projects_df['Funded Month'] = projects_df['Funded Date'].dt.month
projects_df['Funded Day'] = projects_df['Funded Date'].dt.weekday
projects_df['Funded Quarter'] = projects_df['Funded Date'].dt.quarter

# Tiempo transcurrido entre la publicación y el financiamiento del proyecto
projects_df['Funding Time'] = projects_df['Funded Date'] - projects_df['Posted Date']


# =========================
# Categorías de los proyectos
# =========================

t = projects_df['Project Subject Category Tree'].value_counts()

x = list(t.index)
y = list(t.values)

r = {}

for i, val in enumerate(x):
    for each in val.split(","):
        categoria = each.strip()

        if categoria not in r:
            r[categoria] = y[i]

        r[categoria] += y[i]

sorted_x = list(sorted(r.items(), key=operator.itemgetter(1), reverse=True))[:10]

x1 = [a[0] for a in sorted_x][::-1]
y1 = [a[1] for a in sorted_x][::-1]


# =========================
# Subcategorías de los proyectos
# =========================

t1 = projects_df['Project Subject Subcategory Tree'].value_counts()

x2 = list(t1.index)
y2 = list(t1.values)

r = {}

for i, val in enumerate(x2):
    for each in val.split(","):
        subcategoria = each.strip()

        if subcategoria not in r:
            r[subcategoria] = y2[i]

        r[subcategoria] += y2[i]

sorted_x = list(sorted(r.items(), key=operator.itemgetter(1), reverse=True))[:10]

x2 = [a[0] for a in sorted_x][::-1]
y2 = [a[1] for a in sorted_x][::-1]


# =========================
# Categorías de recursos
# =========================

t = projects_df['Project Resource Category'].value_counts()

x3 = list(t.index[::-1])[:10]
y3 = list(t.values[::-1])[:10]


# =========================
# Gráficas
# =========================

trace1 = bar_hor_noagg(
    x1,
    y1,
    "",
    "#8cf2a9",
    600,
    400,
    10,
    rt=True
)

trace2 = bar_hor_noagg(
    x2,
    y2,
    "",
    "#8cf2a9",
    600,
    400,
    10,
    rt=True
)

trace3 = bar_hor_noagg(
    x3,
    y3,
    "",
    "#8cf2a9",
    600,
    400,
    10,
    rt=True
)

fig = tools.make_subplots(
    rows=1,
    cols=5,
    print_grid=False,
    subplot_titles=[
        "Categorías de Proyectos",
        "",
        "Subcategorías de Proyectos",
        "",
        "Categorías de Recursos"
    ]
)

fig.append_trace(trace1, 1, 1)
fig.append_trace(trace2, 1, 3)
fig.append_trace(trace3, 1, 5)

fig['layout'].update(
    height=400,
    showlegend=False,
    margin=dict(l=120)
)

iplot(fig)
```

### 8.1 Limpieza de Texto

Escriba una función de utilidad que normalice el texto y elimine el ruido. Esta función de limpieza de texto realiza la eliminación de palabras vacías (stopwords) del texto y la eliminación de signos de puntuación.

```python
# ==============================
# IMPORTAR LIBRERÍAS
# ==============================

import string
from nltk.corpus import stopwords

# ==============================
# LISTA DE STOPWORDS
# ==============================

stop_words = set(stopwords.words("english"))

# ==============================
# FUNCIÓN PARA LIMPIAR TEXTO
# ==============================

def _clean(txt):

    # Si el valor está vacío
    if txt is None:
        return ""

    # Convertir a texto
    txt = str(txt)

    # Convertir a minúsculas
    txt = txt.lower()

    # Eliminar signos de puntuación
    txt = ''.join(c for c in txt if c not in string.punctuation)

    # Eliminar stopwords
    palabras = []

    for word in txt.split():
        if word not in stop_words:
            palabras.append(word)

    clean_txt = " ".join(palabras)

    # Palabras consideradas ruido
    ruido = [
        "students",
        "student",
        "my",
        "title",
        "would",
        "every",
        "will",
        "many",
        "also",
        "donotremoveessaydivider"
    ]

    for palabra in ruido:
        clean_txt = clean_txt.replace(palabra, "")

    return clean_txt.strip()

# ==============================
# UNIR DATAFRAMES
# ==============================

proyectos = projects_df.merge(
    schools_df,
    on="School ID",
    how="inner"
)

# ==============================
# FILTRAR PROYECTOS DE ABRIL 2018
# ==============================

abril2018_df = proyectos[
    (proyectos["Posted Year"] == 2018) &
    (proyectos["Posted Month"] == 4)
].copy()

# ==============================
# LIMPIAR LOS TEXTOS
# ==============================

abril2018_df["clean_essay"] = (
    abril2018_df["Project Essay"]
    .fillna("")
    .apply(_clean)
)

abril2018_df["clean_need"] = (
    abril2018_df["Project Need Statement"]
    .fillna("")
    .apply(_clean)
)

# ==============================
# MOSTRAR RESULTADOS
# ==============================

abril2018_df[
    [
        "Project ID",
        "clean_essay",
        "clean_need"
    ]
].head()
```

```python
import pandas as pd
import plotly.graph_objs as go
from plotly.subplots import make_subplots

# Contar el número de donaciones por proyecto
num_donaciones = donations_df.groupby('Project ID').agg({'Donation ID' : 'count'}).reset_index().rename(columns={'Donation ID' : 'NumDonaciones'})
df_temporal = projects_df.merge(num_donaciones, on="Project ID")

# Filtrar proyectos de 2017 con 1 solo donante y con más de 10 donantes
df1 = df_temporal[df_temporal['Posted Year'] == 2017][df_temporal['NumDonaciones'] == 1] 
df2 = df_temporal[df_temporal['Posted Year'] == 2017][df_temporal['NumDonaciones'] > 10] 

# Si df2 está vacío, usar umbral más bajo
if len(df2) == 0:
    print("Ajustando umbral a > 3 donantes...")
    df2 = df_temporal[df_temporal['Posted Year'] == 2017][df_temporal['NumDonaciones'] > 3]

print(f"df1: {len(df1)}, df2: {len(df2)}")

# Unir con la información de los profesores usando LEFT JOIN para no perder datos
df1 = df1.merge(teachers_df, on="Teacher ID", how='left')
df2 = df2.merge(teachers_df, on="Teacher ID", how='left')

print(f"Después del merge - df1: {len(df1)}, df2: {len(df2)}")

def obtener_traza(col, func, bandera=None, leyenda=False):
    t1 = df1[col].value_counts()
    t2 = df2[col].value_counts()

    if bandera == 'num':
        t1 = df1[df1['NumProyectos'] <= 20][col].value_counts()    
        t2 = df2[df2['NumProyectos'] <= 20][col].value_counts() 
    
    traza1 = func(
        x=t1.index,
        y=[float(a)*100/sum(t1.values) for a in t1.values],
        name='Donante Único',
        marker=dict(color='#0de286'),
        showlegend=leyenda,
        opacity=0.9
    )
    traza2 = func(
        x=t2.index,
        y=[float(a)*100/sum(t2.values) for a in t2.values],
        name='Múltiples Donantes',
        marker=dict(color='#e82041'),
        opacity=0.9, 
        showlegend=leyenda
    )
    return traza1, traza2

tr1, tr2 = obtener_traza('Project Grade Level Category', go.Scatter)
tr3, tr4 = obtener_traza('School Metro Type', go.Scatter)
tr5, tr6 = obtener_traza('School State', go.Bar)
tr7, tr8 = obtener_traza('Project Resource Category', go.Bar)
tr9, tr10 = obtener_traza('Posted Month', go.Bar, leyenda=True)


########################
# Análisis de longitud del ensayo (conteo de palabras)
df1['wc'] = df1['Project Essay'].apply(lambda x : len(x.split()))
df2['wc'] = df2['Project Essay'].apply(lambda x : len(x.split()))

intervalos = [0, 50, 100, 150, 200, 250, 300, 350, 400, 450, 500, 10000]

df1['wc'] = pd.cut(df1['wc'], intervalos)
t1 = dict(df1['wc'].value_counts())
etiquetas = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t1.keys()]
vals = [float(a)*100/sum(t1.values()) for a in list(t1.values())]
dicc = {}
for i, v in enumerate(etiquetas):
    dicc[v] = vals[i]
orden = ['0-50', '50-100', '100-150', '150-200', '200-250', '250-300', '300-350', '350-400', '400-450', '450-500', '500-10000']
valores = [dicc.get(v, 0) for v in orden]

tr13 = go.Bar(
    x=orden,
    y=valores,
    name='Donante Único',
    marker=dict(color='#0de286'),
    showlegend=False,
    opacity=0.9
)


df2['wc'] = pd.cut(df2['wc'], intervalos)
t2 = dict(df2['wc'].value_counts())
etiquetas2 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t2.keys()]
vals2 = [float(a)*100/sum(t2.values()) for a in list(t2.values())]
dicc2 = {}
for i, v in enumerate(etiquetas2):
    dicc2[v] = vals2[i]
valores2 = [dicc2.get(v, 0) for v in orden]

tr14 = go.Bar(
    x=orden,
    y=valores2,
    name='Múltiples Donantes',
    marker=dict(color='#e82041'),
    showlegend=False,
    opacity=0.9
)

############### costo ###########
intervalos = [0, 100, 500, 1000, 1500, 2000, 2500, 3000, 70000]

df1['binnedcost'] = pd.cut(df1['cost'], intervalos)
t1 = dict(df1['binnedcost'].value_counts())
etiquetas = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t1.keys()]
vals = [float(a)*100/sum(t1.values()) for a in list(t1.values())]
dicc = {}
for i, v in enumerate(etiquetas):
    dicc[v] = vals[i]
orden_costo = ['100-500', '500-1000', '1000-1500', '1500-2000', '2000-2500', '2500-3000', '3000-70000']
valores = [dicc.get(v, 0) for v in orden_costo]

tr15 = go.Scatter(
    x=orden_costo,
    y=valores,
    name='Donante Único',
    fill='tozeroy',
    marker=dict(color='#0de286'),
    showlegend=False,
    opacity=0.9
)


df2['binnedcost'] = pd.cut(df2['cost'], intervalos)
t2 = dict(df2['binnedcost'].value_counts())
etiquetas2 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t2.keys()]
vals2 = [float(a)*100/sum(t2.values()) for a in list(t2.values())]
dicc2 = {}
for i, v in enumerate(etiquetas2):
    dicc2[v] = vals2[i]
valores2 = [dicc2.get(v, 0) for v in orden_costo]

tr16 = go.Scatter(
    x=orden_costo,
    y=valores2,
    name='Múltiples Donantes',
    fill='tozeroy',
    marker=dict(color='#e82041'),
    showlegend=False,
    opacity=0.9
)

#############
titulos = [
    'Nivel de Grado del Proyecto', 
    'Tipo de Zona Escolar', 
    'Costo del Proyecto ($)', 
    'Cantidad de Palabras del Ensayo', 
    'Mes de Publicación', 
    'Categoría de Recurso'
]

fig = make_subplots(
    rows=3, cols=2, 
    subplot_titles=titulos, 
    specs=[[{}, {}], [{}, {}], [{}, {}]]
) 

fig.add_trace(tr1, row=1, col=1)
fig.add_trace(tr2, row=1, col=1)
fig.add_trace(tr3, row=1, col=2)
fig.add_trace(tr4, row=1, col=2)
fig.add_trace(tr15, row=2, col=1)
fig.add_trace(tr16, row=2, col=1)
fig.add_trace(tr13, row=2, col=2)
fig.add_trace(tr14, row=2, col=2)
fig.add_trace(tr7, row=3, col=2)
fig.add_trace(tr8, row=3, col=2)
fig.add_trace(tr9, row=3, col=1)
fig.add_trace(tr10, row=3, col=1)

fig.update_layout(
    barmode='group',
    title=dict(
        text='<b>Comparación de Proyectos: Donante Único vs Múltiples Donantes</b>',
        font=dict(size=22, color='#000')
    ),
    legend=dict(x=0.25, y=1.02, orientation="h"),
    margin=dict(t=120, b=100),
    paper_bgcolor='rgb(254, 247, 234)',
    plot_bgcolor='rgb(254, 247, 234)',
    height=1000,
    showlegend=True
)

fig.show()
```
